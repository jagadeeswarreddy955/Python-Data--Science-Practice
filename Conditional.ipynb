#Conditional Statements


If Statement

```python
age=20
if age>=18:
  print("you are eligible to vote")
```

if-else

```python
num=7
if num%2==0:
  print("Even Number")
else:
  print("Odd Number")
```

if-elif-else

```python
marks=int(input("enter the marks: "))
if marks>=90:
    print("Grade A")
elif marks>=80:
    print("Grade B")
elif marks>=70:
    print("Grade C")
elif marks>=60:
    print("Grade D")
else:
    print("Grade F")
```

#Examples on Conditional Statements


1.Check weather a number is positive or negative

```python
num=int(input("Enter a number: "))
if num<0:
  print("Negative Number")
elif num>0:
  print("Positive Number")
else:
  print("Whole Number")
```

Check wheather a person passed or fail...

```python
marks=25
if marks >=35:
  print("Pass")
else:
  print("Fail")
```
