#Functions

Function is a block of code.Run when its called.

```python
def func():
  print("this is function")
func()
```

Functions with parameters and arguments

```python
def sum(a,b):#defination of a function & a,b are parameters
  return a+b  #function body
print(sum(1,2))  #function call & 1,2 are aurgments
```

#Nested Function
Function in function is called Nested Function

```python
def outer():
  print("outer function")
  def inner():
    print("inner function")
  inner()
outer()
```

#Orbitary arguments in fumctions
multiple arguments through one parameter by using '*' is called Orbitary arguments

```python
def func(*a):
  print("this is function",a)
func(1,2,3,4)
```

#Functions with key word arguments

```python
def func(**b):
  print("this is function",b)
func(a=2,b=3)
```

#Example on Functions
Check wheather the numberr is even or odd

```python
def check_number(n):
  if n%2==0:
    print("The number is even")
  else:
    print("The number is odd")
check_number(10)
check_number(9)
```

Function to find a square of a number

```python
def square(n):
  return n*n
print(square(2))
print(square(8))
```

Function to find a largest of two numbers

```python
def largest(a,b):
  if a>b:
    print("A is largest")
  else:
    print("B is largest")
largest(1,2)
```

Functions to calculate Factorial

```python
def function(n):
  result=1
  for i in range(1,n+1):
    result=result*i
  return result
print(function(5))
```

Functions to calculate total marks and aveerage marks

```python
def student_marks(m1,m2,m3):
  total_marks=m1+m2+m3
  average_marks=total_marks/3
  print("total marks",total_marks)
  print("Average marks",average_marks)
student_marks(30,40,50)
```
