#Tuples()
Ordered,immutable,allow duplicates,allow indexing and slicing

Tuples allows only index and slicing

```python
t=(1,2,3,4,5,"jaggu")
print(t)
print(type(t))
print(t[1])
print(t[0:4:2])
```

#Tuple operations
Concatination--Adding of two tuples

```python
t1=(1,2)
t2=(2,3)
print(t1+t2)
```

Iteration

```python
t=(1,2,3,2,5,"jaggu")
for i in t:
  print(i)
```

Repetation

```python
t=(1,2,3,2,5,"jaggu")
print(t*5)
```

Membership operators

```python
t=(1,2,3,2,5,"jaggu")
t3=(1,2,3,4,5,)
print(1 in t)
print(1 not in t)

print(t is t3)
print(t is not t3)
```
