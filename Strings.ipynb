#Strings

A string is a sequence of characters which are in quotes (' ') or double quotes(" ") or triple quotes('''  ''').

```python
name = "Jagadeesh"
print(name)
```

Length of string

```python
name = "Python"
print(len(name))
```

Upper case()

```python
name = "python"
print(name.upper())
```

Lower case()

```python
name = "PYTHON"
print(name.lower())
```

Indexing

```python
word = "Python"
print(word[0])
print(word[2])
print(word[-1])
```

Slice a string

```python
word = "Python"
print(word[0:3])
print(word[2:5])
```

Reverse a string

```python
word = "Python"
print(word[::-1])
```

Replace characters or words

```python
text = "I like Java"
print(text.replace("Java", "Python"))
```

Remove extra spaces(strip)

```python
text = "   Hello Python   "
print(text.strip())
```

Split a string into a list

```python
text = "Python SQL PowerBI"
print(text.split())
```

capitalize()

```python
text = "hello PYTHON"
print(text.capitalize())
```

title()

```python
text = "learn python programming"
print(text.title())
```

swapcase()

```python
text = "PyThOn"
print(text.swapcase())
```

#Searching Methods
find()

```python
text = "Hello Python"
print(text.find("Python"))
print(text.find("Java"))
```

rfind()

```python
text = "apple banana apple"
print(text.rfind("apple"))
```

count()

```python
text = "banana"
print(text.count("a"))
```

startswith()

```python
text = "Python programming"
print(text.startswith("Python"))
```

endswith()

```python
filename = "report.pdf"
print(filename.endswith(".pdf"))
```
