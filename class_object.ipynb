#Class and Objects

```python
class students:
  name="ravi"
  salary=25000
s1=students()
print(s1.name)
print(s1.salary)
```

```python
class student:
  name="jaggu"
  marks=80
s1=student()
print(s1.name)
print(s1.marks)
```

#__init__
init() is  method that python calls automatically when you create an object.

```python
class student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks
s1=student("ravi",68)
print(s1.name)
print(s1.marks)

```

#Self()
Self refers to current object.You can access the objects attributes and methods.

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Ravi")
s2 = Student("Priya")

print(s1.name)
print(s2.name)

```

```python
class student:
  def __init__(self,name,salary,age):
    self.name=name
    self.salary=salary
    self.age=age
  def display(self):
    print("Name",self.name)
    print("salary",self.salary)
    print("Age",self.age)
e1=student("ravi",25000,30)
e2=student("priya",30000,28)
e3=student("jaggu",40000,23)
e1.display()
e2.display()
e3.display()
```

```python
class Product:
    category = "Electronics"

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display(self):
        print("Name:", self.name)
        print("Price:", self.price)
        print("Category:", self.category)

p1 = Product("Laptop", 50000)
p1.display()
```
