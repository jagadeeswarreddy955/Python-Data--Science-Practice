#File Handling

Read mode


```python
s=open('/content/demo.txt',mode='r')
print(s.read())
s.close()
```

Write mode

```python
s=open('/content/demo.txt',mode='w')
s.write("bye bye")
s.close()
```

Append mode


```python
s=open('/content/demo.txt',mode='a')
s.write("byee byee write")
s.close()
```

'r+' mode both read and write

```python
s=open('/content/demo.txt','r+')
print(s.read())
s.write("r+ mode")
s.close()
```

```python
s=open('/content/demo.txt','w+')
s.write("w+ mode")
s.seek(0)
print(s.read())
s.close()
```
