#Loops

#For loop

Print numbers from 1 to 5

```python
for i in range(1, 6):
    print(i)
```

Print even numbers from 2 to 10

```python
for i in range(2, 11, 2):
    print(i)
```

Sum of numbers from 1 to 5

```python
total = 0

for i in range(1, 6):
    total = total + i

print("Total:", total)
```

Multiplication table

```python
num = 5

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

#While Loop

Print numbers from 1 to 5

```python
i = 1

while i <= 5:
    print(i)
    i = i + 1
```

Print numbers from 10 to 1

```python
i = 10

while i >= 1:
    print(i)
    i = i - 1
```

Sum of digits using a while loop

```python
num = 1234
total = 0

while num > 0:
    digit = num % 10
    total = total + digit
    num = num // 10

print("Sum:", total)
```

#Loop control statements

Break

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

Continue

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```
