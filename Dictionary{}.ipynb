#Dictionatry{}
Key value pairs,Key should be immutable,Value should be immutable,No slicing,Keys are unique

get()

```python
a={1:"abc","fgj":4,"python life":6}
print(a.get(1))
```

update()

```python
a={1:"abc","fgj":4,"python life":6}
a.update({1:"xyz"})
print(a)
```

values()

```python
a={1:"abc","fgj":4,"python life":6}
print(a.values())
```

keys()

```python
a={1:"abc","fgj":4,"python life":6}
print(a.keys())
```

items()

```python
a={1:"abc","fgj":4,"python life":6}
for i in a.items():
    print(i)
```
