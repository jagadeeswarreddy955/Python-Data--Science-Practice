#Exception Handling

The way handle errors during program execution

```python
try:
  print(a)
except:
  print("error")
else:
  print("no error")
finally:
  print("always")
```

try-except

```python
try:
  a=10
  b=0
  print(a/b)
except:
  print("cannot divode by zero")
```

Value Error

```python
try:
  a=int(input("enter the age:"))
  print("your age is",a)
except ValueError:
  print("enter valid age")
```

FileNotFoundError

```python
try:
  with open("file.txt",'r') as f:
    print(f.read())
except FileNotFoundError:
  print("file not found")
```

We can use multiple except blocks to handle different types of errors.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    print(a / b)

except ValueError:
    print("Please enter numbers only")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

Using else with exception handling

```python
try:
    a = 10
    b = 2
    result = a / b

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Result:", result)
```

Using finally

```python
try:
    print(10 / 2)

except ZeroDivisionError:
    print("Division error")

finally:
    print("Program completed")
```

Complete example using try, except, else, and finally

```python
try:
    num = int(input("Enter a number: "))
    result = 100 / num

except ValueError:
    print("Enter a valid integer")

except ZeroDivisionError:
    print("Number cannot be zero")

else:
    print("Result:", result)
```
