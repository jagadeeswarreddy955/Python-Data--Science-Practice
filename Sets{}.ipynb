#Sets{}
Do not duplicates,No index,Unordered,Do not allow mutable types as set elements

add()


```python
a={1,2,3,4,5,6}
a.add(7)
print(a)
```

update()

```python
a={1,2,3,4,5,6}
a.update([7,8,9])
print(a)
```

pop()

```python
a={1,2,3,4,5,6}
a.pop()
print(a)
```

remove()

```python
a={1,2,3,4,5,6}
a.remove(2)
print(a)
```

Union()

```python
set1={1,2,3,4,5,6,7}
set2={1,2,3}
print(set1.union(set2))
```

Intersection()


```python
set1={1,2,3,4,5,6,7}
set2={1,2,3}
print(set1.intersection(set2))
```

Difference()

```python
set1={1,2,3,4,5,6,7}
set2={1,2,3}
print(set1.difference(set2))
```

issubset()

```python
set1={1,2,3,4,5,6,7}
set2={1,2,3}
print(set2.issubset(set1))
```

issuperset()

```python
set1={1,2,3,4,5,6,7}
set2={1,2,3}
print(set1.issuperset(set2))
```
