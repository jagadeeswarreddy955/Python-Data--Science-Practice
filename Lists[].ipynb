#Lists
[],
Mutable data type,ordered,
allow duplicates,lists maintain insertion order

Append

```python
a=[1,2,3,4.5,"kiran"]
a.append("jaggu")
print(a)
```

Extend

```python
a=[1,2,3,4.5,"kiran"]
a.extend([6,7])
print(a)
```

Remove

```python
a=[1,2,3,4.5,"kiran"]
a.remove(3)
print(a)
```

Pop

```python
a=[1,2,3,4.5,"kiran"]
a.pop(0)
print(a)
```

Count

```python
a=[1,2,3,4.5,1,"kiran",1]
print(a.count(1))
```

Index

```python
a=[1,2,3,4.5,"kiran"]
print(a.index("kiran"))
```

```python
a=[1,2,3,4.5,"kiran"]
for i in a:
  print(i)
```
